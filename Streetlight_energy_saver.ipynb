```python
import random   # Used to simulate sensor movement detection

# ---- Step 1: Define streetlights and their base settings ----
streetlights = [
    {"id": 1, "brightness": 30},   # Streetlight 1 starts at 30% brightness
    {"id": 2, "brightness": 30},   # Streetlight 2 starts at 30%
    {"id": 3, "brightness": 30},   # Streetlight 3 starts at 30%
]

# ---- Step 2: Function that simulates sensor movement detection ----
def detect_movement():
    # Randomly return True/False to imitate movement presence
    return random.choice([True, False])

# ---- Step 3: Track energy usage before and after optimization ----
baseline_energy = 0      # Total energy if lights stayed bright all the time
optimized_energy = 0     # Total energy after adaptive brightness control

# ---- Step 4: Adaptive brightness logic for each streetlight ----
for light in streetlights:                          # Loop through each streetlight
    movement = detect_movement()                    # Check if movement is detected

    if movement:
        light["brightness"] = 100                   # Increase brightness to 100%
    else:
        light["brightness"] = 20                    # Dim to 20% to save energy

    baseline_energy += 100                          # Baseline assumes always 100% power
    optimized_energy += light["brightness"]         # Add the adaptive brightness value

# ---- Step 5: Compute total energy saved ----
energy_saved = baseline_energy - optimized_energy   # Difference between constant vs adaptive use

# ---- Step 6: Display results ----
print("Final Streetlight Brightness Levels:", streetlights)  # Show brightness decisions
print("Total Energy Saved:", energy_saved, "%")              # Energy saved metric
```
