```python
# -----------------------------------------------
# Pollution Control Monitors : Grid Simulation
# -----------------------------------------------

import numpy as np               # For numerical grid simulation
import matplotlib.pyplot as plt  # For plotting pollution over time

# -----------------------------
# 1. Initialize Simulation Grid
# -----------------------------

np.random.seed(0)   # Fix randomness for reproducible results

grid_size = 10      # Create a 10x10 grid to represent city zones
timesteps = 50      # Number of simulation iterations

# Create initial pollution levels (0–50 range)
pollution_grid = np.random.uniform(0, 50, (grid_size, grid_size))

# List to store total pollution in each timestep
pollution_over_time = []

# -----------------------------
# 2. Agent-Based Simulation
# -----------------------------

for t in range(timesteps):     # Loop over each timestep in simulation

    # Store total pollution of this timestep
    pollution_over_time.append(np.mean(pollution_grid))

    # Each agent senses pollution in its local area (3×3 neighborhood)
    new_grid = pollution_grid.copy()

    for i in range(grid_size):
        for j in range(grid_size):

            # Extract local 3×3 zone around (i,j)
            i_min, i_max = max(0, i-1), min(grid_size, i+2)
            j_min, j_max = max(0, j-1), min(grid_size, j+2)

            local_zone = pollution_grid[i_min:i_max, j_min:j_max]

            local_avg = np.mean(local_zone)      # Calculate local average pollution

            # If pollution exceeds threshold → mitigation starts
            if local_avg > 40:
                new_grid[i, j] -= 2    # Agent reduces pollution
            elif local_avg < 15:
                new_grid[i, j] += 1    # Pollution drifts slightly (natural increase)

    # Update grid after all agents act
    pollution_grid = np.clip(new_grid, 0, 100)  # Keep values within valid range

# -----------------------------
# 3. Graph Output
# -----------------------------

plt.plot(pollution_over_time)           # Plot total pollution trend
plt.title("Pollution Levels Over Time") # Title of the graph
plt.xlabel("Time Step")                 # X-axis label
plt.ylabel("Average Pollution")         # Y-axis label
plt.grid(True)                          # Add gridlines
plt.show()                              # Display the graph
```
